## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

```python
import os
import pandas as pd
from google.colab import userdata
from datasets import load_dataset

os.environ["HF_TOKEN"] = userdata.get("HF_TOKEN")
dataset = load_dataset("FlyRank/internship-warehouse", "fact_content_daily_performance", split="train", streaming=True)

for row in dataset.take(1):
    print("Available columns:", list(row.keys()))

data_rows = []
for idx, row in enumerate(dataset):
    data_rows.append(row)
    if idx >= 1500:
        break

df = pd.DataFrame(data_rows)


numeric_cols = df.select_dtypes(include=['number']).columns.tolist()
print("Numeric columns found:", numeric_cols)

if len(numeric_cols) >= 2:
    df['score'] = df[numeric_cols[0]].fillna(0) * 2.0 + df[numeric_cols[1]].fillna(0) * 0.5
else:
    df['score'] = 1.0

df['reason_code'] = 'BASELINE_PRIORITY'
df['action'] = 'Review Content'

df_ranked = df.sort_values(by='score', ascending=False).reset_index(drop=True)

os.makedirs('work/outputs', exist_ok=True)
df_ranked.head(20).to_csv('work/outputs/baseline_action_score.csv', index=False)

print("Successfully generated and saved work/outputs/baseline_action_score.csv!")
```
