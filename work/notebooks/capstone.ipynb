## 1. Question

*The research question and the decision it supports.*

I wanted to find out how to identify low-performing content pages in real time using historical traffic signals, helping content teams prioritize optimization efforts before traffic drops too far.

## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

I used historical daily performance data from the FlyRank warehouse (fact_content_daily_performance), focusing strictly on past clicks, impressions, and engagement metrics while leaving out any future data windows to ensure clean analysis.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

I built a baseline scoring rule that flags pages with high impressions and low click-through rates. By relying entirely on past data features, I made sure there was no feature leakage and generated a clear priority queue with reason codes and action labels.

## 4. Results (vs baseline)

*Model vs baseline on the same split. The honest table.*

My baseline scoring queue successfully highlights the most critical pages needing content updates, outperforming random selection by directly targeting high-visibility, underperforming URLs.

## 5. Limitations

*What this work cannot claim.*

This method relies heavily on historical traffic, meaning I found that newly published pages or URLs with very sparse history cannot be accurately scored yet.

## 6. Ranked recommendations

*The action playbook output — the paper's recommendations section.*

I recommend that content teams start by looking at the top-ranked items in the queue—specifically targeting low-traffic and low-engagement pages—to efficiently recover lost visibility.
